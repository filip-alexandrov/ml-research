```python
# tensor = mulitdimensional array
import torch 

print("Scalar value, uninit: ", torch.empty(1))

print("Vector value: ", torch.empty(3))

print("2D matrix: ", torch.empty(2, 3))

print("3D matrix: ", torch.empty(2, 2, 3))
```

```python
# 2 rows, 3 columns
torch.rand(2, 3)
```

```python
# 2 matrices (z), 3 rows, 4 columns, like 2 conv filters
torch.rand(2, 3, 4)
```

```python
torch.zeros(2, 3)
```

```python
torch.ones(2, 3)
```

```python
x = torch.ones(2, 2, dtype=torch.float64) # double, float16
print(x.dtype)
print(x.size())
```

```python
x = torch.tensor([2.5, 0.1]) # tensor from list of values
x.dtype
```

```python
x = torch.randint(high = 10, size = (2, 2)) 
y = torch.randint(high = 10, size = (2, 2)) 

x, y
```

```python
# elementwise addtion 
x + y
torch.add(x, y) 
```

```python
# inplace addition, modifies y
y.add_(x)
```

```python
x - y 
torch.sub(x, y) 
y.sub_(x)
```

```python
torch.mul(x, y) 
y.mul_(x) 
```

```python
# elementwise division
x / y 
torch.div(x, y)
```

```python
x = torch.rand(5, 3)

# slicing = all rows, but only first column
print(x[: , 0 ])

# row 1, all cols
print(x[1, :])
```

```python
# only when you have one element tensor
x[1,1].item() # actual value (out of tensor)
```

```python
x = torch.rand(4, 4) 
print(x)

# reshape tensor, extends rows with each other
x.view(16)
```

```python
# auto determine size of -1 dimension
x.view(-1, 8)
```

```python
import numpy as np 

a = torch.ones(5) 

# if on the cpu, both share same pointer
b = a.numpy()

# broadcasted to all elements
a.add_(1)

b
```

```python
a = np.ones(5)

b = torch.from_numpy(a)
b.to(dtype=torch.float16)

# points to same memory 
b.add_(1)
```

```python
device = torch.device("mps")
x = torch.ones(2, 2, device=device)
y = torch.ones(2, 2, device=device)

z = x + y
z
```

```python
# move back to cpu 
z = z.to("cpu")
```
