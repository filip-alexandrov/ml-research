```python
import torch 

# calculate grandients with respect to x 
x = torch.randn(3, requires_grad=True) 
x
```

```python
y = x + 2
y
```

```python
# multipy backward function for gradient calculations
z = y * y * 2 
z
```

```python
# mean backward function
z = z.mean()
z
```

```python
# calculate gradients 
z.backward() # dz/dx

# grad can only be implicitly created for scalar outputs
x.grad
```

```python
x = torch.randn(3, requires_grad=True)
y = x + 2
z = y * y * 2

# v is the weights for grad calculation (from loss function)
v = torch.tensor([0.1, 1.0, 0.0001], dtype=torch.float32)
z.backward(v)
x.grad
```

```python
# remove grad tracking 
# x.requires_grad_(False)
# x

# new tensor with no grad tracking
# x.detach()
# x

with torch.no_grad(): 
    y = x + 2

    # no grad_fn for y
    print(y)
```

```python
weights = torch.ones(4, requires_grad=True) 

for epoch in range(3): 
    model_output = (weights * 3).sum()
    model_output.backward()

    # accumulates gradients => 3 * i
    print(weights.grad)

    # empty gradients before next step 
    weights.grad.zero_()
```

```python
optimizer = torch.optim.SGD([weights], lr = 0.01) 
optimizer.step()
optimizer.zero_grad()
```

```python
# random input tensor
x = torch.randn(3, requires_grad=True) 

# simple model operation, linear transformation without bias
W = torch.randn(3, 3) # weight matrix 
z = torch.matmul(W, x) # Wx predicted output, still a vector 

# true values for each of the predictions 
y_true = torch.randn(3) 

# compute the mean squared error loss
loss = (z - y_true) ** 2 
mse_loss = torch.mean(loss) 




```

```python
# backpropagate the loss (automatic)
mse_loss.backward()

# now x.grad will contain the gradients of mse_loss with respect to x 
print("Gradient of x: ", x.grad) 
```

```python
# If you want to use z.backward(v) you'd compute v as the gradient of mse_loss with respect to z
# Since mse_loss = 1/n * sum((z_i - y_true_i)^2), the partial derivative with respect to z_i is:
v = 2.0 * (z - y_true) / len(z)

# backward in manual mode, v is 
z.backward(v)

print("Gradient of x using manual v: ", x.grad)
```
