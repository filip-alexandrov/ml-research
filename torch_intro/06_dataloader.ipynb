```python
# 1 epoch = 1 forward and backward pass of ALL training samples 

# batch_size = number of training samples in one forward & one backward pass

# number of iterations = number of passes, each pass using [batch_size] number of samples 

# eg. 100 samples, batch_size = 20 => 100/20 = 5 iterations for 1 epoch
```

```python
import torch 
import torchvision
from torch.utils.data import Dataset, DataLoader
import numpy as np 
import math 
```

```python
class WineDataset(Dataset): 
    def __init__(self) -> None:
        # should load the data
        xy = np.loadtxt("data/wine.csv", delimiter=",", dtype=np.float32, skiprows=1)

        # split into X and y
        self.x = torch.from_numpy(xy[:, 1:]) # all rows, second column to last column, (n_samples, n_features)
        self.y = torch.from_numpy(xy[:, [0]]) # all rows, first column (n_samples, 1)
        self.n_samples = xy.shape[0]

        # dataloading 
    
    def __getitem__(self, index) -> tuple:
        # will allow dataset[0]
        return self.x[index], self.y[index]
    
    def __len__(self) -> int:
        # will allow len(dataset) 
        return self.n_samples

dataset = WineDataset()
```

```python
features, labels = dataset[0]

features, labels
```

```python
# num_workers > 0 uses multiple cores for loading, cannot be used in jupyter, should be in __main__
dataloader = DataLoader(dataset=dataset, batch_size=4, shuffle=True)

dataiter = iter(dataloader) 
data = dataiter._next_data()

# batch size = 4 => 4 feature vectors, 4 labels
features, labels = data
features, labels
```

```python
# training loop 
num_epochs = 2 
total_samples = len(dataset) 
n_iterations = math.ceil(total_samples / 4 )

f"Total samples: {total_samples}, Total iterations: {n_iterations}"
```

```python
for epoch in range(num_epochs): 
    for i, (inputs, labels) in enumerate(dataloader): 
        # forward, backward, update 
        if (i + 1) % 5 == 0: 
            print(f"epoch {epoch + 1}/{num_epochs}, step {i + 1}/{n_iterations}, inputs {inputs.shape}")
```
