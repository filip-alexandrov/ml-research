```python
import torch 
from torch.utils.data import Dataset
import torchvision
import numpy as np
```

```python
# on images : crop, center, grayscale, padding
# on tensors : linear transform, normalize, random erasing 
# generic : lambdas, custom class 
# multiple transforms : compose 

from typing import Any


class WineDataset(Dataset): 
    def __init__(self, transform = None) -> None:
        # should load the data
        xy = np.loadtxt("data/wine.csv", delimiter=",", dtype=np.float32, skiprows=1)

        # split into X and y
        self.x = xy[:, 1:] # all rows, second column to last column, (n_samples, n_features)
        self.y = xy[:, [0]] # all rows, first column (n_samples, 1)

        self.transform = transform

    def __getitem__(self, index) -> tuple:
        # will allow dataset[0]
        sample = self.x[index], self.y[index]

        if self.transform: 
            sample = self.transform(sample)

        return sample 
    
    def __len__(self) -> int:
        # will allow len(dataset) 
        return self.n_samples
    
class ToTensor: 
    def __call__(self, sample):
        inputs, targets = sample 
        return torch.from_numpy(inputs), torch.from_numpy(targets)
    
class MulTransform: 
    def __init__(self, factor):
        self.factor = factor
        
    def __call__(self, sample): 
        inputs, target = sample 
        inputs *= self.factor 
        return inputs, target 
        

composed = torchvision.transforms.Compose([ToTensor(), MulTransform(4)])

# dataset = WineDataset(transform=ToTensor())
dataset = WineDataset(transform=composed)

features, labels = dataset[0]
features, labels
```
