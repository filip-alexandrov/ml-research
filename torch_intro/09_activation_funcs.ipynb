activation funcs have to be non-linear <br>
step func: not used in practice (non-differentiable, info loss) <br>
sigmoid func: outputs probability between 0 - 1 (typically used in last layer of binary classifier) <br>
tanh: scaled & shifted sigmoid, outputs -1 to +1 (used in hidden layers - centered around 0, avoids vanishing/exploding gradients) <br>
relu: most popular in practice for hidden layers (like linear - retains a lot of info, but still inserts non-linearity) <br>
leaky relu: improved relu, tries to solve the vanishing grad problem (when x < 0) = dead neurons which will never be updated <br>
softmax: squashes outputs so they will be 0 - 1, but will sum to 1 (probabilities for multiclass classification) <br>

```python
import torch 
import torch.nn as nn
import torch.nn.functional as F

# activations in predefined layers
class NeuralNet(nn.Module): 
    def __init__(self, input_size, hidden_size) -> None:
        super(NeuralNet, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(hidden_size, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.linear1(x)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        return out
    
# activations in forward pass
class NeuralNet2(nn.Module): 
    def __init__(self, input_size, hidden_size) -> None:
        super(NeuralNet2, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size)
        self.linear2 = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out = F.relu(self.linear1(x))
        out = F.sigmoid(self.linear2(out))
        # out = F.leaky_relu(out)
        # out = F.tanh(out)
        return out  
```
