```python
import math 

def e_x(x): 
    return 1 + x + x**2/2 + x**3/6 + x**4/24 + x**5/120 + x**6/720 + x**7/5040 + x**8/40320 + x**9/362880 + x**10/3628800

def e_x_minus_exp(x):
    absolute_error = abs(e_x(x) - math.exp(x))
    relative_error = absolute_error / math.exp(x)
    return absolute_error, relative_error

# for i in range(1 << 100): 

e_x_minus_exp(10)
```

```python
import numpy as np 
import matplotlib.pyplot as plt

xs = np.linspace(1, 3, 1000) 
ys = [e_x_minus_exp(x) for x in xs]


plt.plot(xs, ys)
plt.show()
```

```python
# cordic algorithm 

def cordic(x):
    k = 1
    e_k = math.exp(k)

    x_new = x - k
    e_x = e_k * (1 + x_new + x_new**2/2 + x_new**3/6 + x_new**4/24 + x_new**5/120)

    true = math.exp(x)
    numeric = e_x
    return true, numeric

xs = np.linspace(-1, 3, 1000) 
ys = [cordic(x) for x in xs]

plt.plot(xs, ys)
plt.xlabel('x')
plt.ylabel('out')
plt.title('CORDIC')
plt.legend(['true', 'numeric'])
plt.show()
```
